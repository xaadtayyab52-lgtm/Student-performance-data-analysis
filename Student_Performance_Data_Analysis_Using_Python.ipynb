```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Student": ["Ali", "Ahmed", "Sara", "Hina", "Usman", "Ayesha"],
    "Study_Hours": [2, 4, 5, 3, 6, 7],
    "Attendance": [70, 80, 90, 75, 95, 98],
    "Marks": [55, 65, 80, 60, 85, 92]
}

df = pd.DataFrame(data)

print(df)
```

```python
print(df.info())
```

```python
print(df.describe())
```

```python
plt.scatter(df["Study_Hours"], df["Marks"])

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")

plt.show()
```

```python
plt.scatter(df["Attendance"], df["Marks"])

plt.xlabel("Attendance")
plt.ylabel("Marks")
plt.title("Attendance vs Marks")

plt.show()
```

```python
average_marks = df["Marks"].mean()

print("Average Marks:", average_marks)
```

```python
highest_marks = df["Marks"].max()

print("Highest Marks:", highest_marks)
```

```python
lowest_marks = df["Marks"].min()

print("Lowest Marks:", lowest_marks)
```

Findings / Insights
1. Students who studied more hours generally achieved higher marks.
2. Students with higher attendance generally had better marks.
3. The highest marks were 92.
4. The lowest marks were 55.
5. The average marks were approximately 72.83.
